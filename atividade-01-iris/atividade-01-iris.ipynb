### Configuração do ambiente

```python
# Importação do PyTorch
import torch

# Verificação da versão e disponibilidade da GPU
print("Versão do PyTorch:", torch.__version__)
print("GPU disponível:", torch.cuda.is_available())

# Definição do dispositivo de processamento
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Dispositivo utilizado:", device)
```

### Carregamento da base Íris

```python
# Base de dados Iris
from sklearn.datasets import load_iris
import pandas as pd

# Carregamento da base
iris = load_iris()

# Separação das características (X) e dos rótulos (y)
X = iris.data
y = iris.target

# Nomes das espécies
nomes = iris.target_names

print("Formato de X:", X.shape)
print("Espécies:", nomes)
```

### Conversão para tensores

```python
# Conversão dos dados para tensores PyTorch
X_t = torch.tensor(
    X,
    dtype=torch.float32,
    device=device
)

y_t = torch.tensor(
    y,
    dtype=torch.long,
    device=device
)

print("Formato de X_t:", X_t.shape)
print("Formato de y_t:", y_t.shape)
```

### Seleção e padronização das características

```python
# Seleção das características relacionadas à pétala
# Índices 2 e 3 correspondem a:
# - comprimento da pétala
# - largura da pétala

X2 = X[:, 2:4]

# Cálculo da média e do desvio padrão
media = X2.mean(axis=0)
desvio = X2.std(axis=0)

# Padronização Z-score
# Isso coloca as características em uma escala comparável.
X2 = (X2 - media) / desvio

# Conversão para Tensor
X2_t = torch.tensor(
    X2,
    dtype=torch.float32,
    device=device
)

print("Formato de X2:", X2.shape)
print("Formato de X2_t:", X2_t.shape)
```
